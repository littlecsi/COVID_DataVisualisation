# COVID-19 Data Analysis

## Purpose of the Project

 이 프로젝트의 목표는 코로나 데이터를 사용자가 쉽게 볼 수 있게 시각화 하여 보여주는 것입니다. 또한, 사용자의 입력에 따라 다른 결과를 출력할 수 있도록 하는 것이 이 프로젝트가 지향하는 최종목표입니다.

## Reason for selecting this topic

 이 주제를 선정한 이유는 코로나 때문에 전국의 학생들이 영향을 받았으므로, 코로나의 영향력을 시각화하여 사용자에게 보여주는 기술을 습득하고, 더 나아가, 이를 활용하여 코로나 데이터를 분석하기 위해서 입니다. 

## Data Acquisition

 저는 데이터를 https://www.kaggle.com/datasets 에서 다운 받을 생각입니다.

## Method

데이터셋을 다운받고 난 후에, pandas 라이브러리를 사용하여 데이터를 불러와 저장할 생각입니다. 그리고, numpy 등 다양한 데이터 분석 관련 라이브러리를 활용해 데이터를 정리 및 분석할 것이고, matplotlib 와 tkinter 등을 사용해 시각화를 진행할 생각입니다.
